```python
tsv_text="""student_id\tName\tDepartment\tAttendance\tMarks
101\tAman\tCSE\t92\t85
102\tPriya\tIT\t96\t92
103\tRahul\tCSE\t\t76
104\tSara\tESE\t88\t89
103\tNeha\tIT\t79\t
103\tVikram\tCSE\t91\t81
"""

with open("students_updated.tsv","w",encoding="utf-8") as file:
  file.write(tsv_text)
  print("TSV file created successfully.")
```

# **Handle missing values**

```python
print(df.isna().sum())
```

```python
df_drop=df.dropna()
display(df_drop)
```

```python
df_fill=df.copy()
df_fill["Attendance"]=df_fill["Attendance"].fillna(df_fill["Attendance"].median())
```

```python
df_fill["Marks"]=df_fill["Marks"].fillna(df_fill["Marks"].median())
```

```python
display(df_fill)
```

```python
print("Summart statistics:")
display(df_fill[["Attendance","Marks"]].describe())
```

```python
display(df_fill.groupby("Department")["Marks"].mean())
```

```python
display(df_fill["Department"].value_counts())
```

```python
df_fill.to_csv("students_cleaned.tsv",sep="\t",index=False)
print("cleaned tsv saved")
```
