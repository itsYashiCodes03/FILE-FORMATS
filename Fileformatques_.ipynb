# Q1.Create a TSV file containing at least 10 employees with the columns "EmployeeID", "Name", "Department", "Experience", and "Salary".

```python
import pandas as pd

data = {
    "EmployeeID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Name": ["Aman", "Riya", "Rahul", "Priya", "Karan",
             "Neha", "Arjun", "Sneha", "Vikas", "Anjali"],
    "Department": ["IT", "HR", "IT", "Finance", "HR",
                   "IT", "Finance", "Sales", "Sales", "Finance"],
    "Experience": [5, 3, 8, 6, 2, 4, 10, 7, 3, 9],
    "Salary": [50000, 35000, 75000, 60000, 30000,
               45000, 80000, 42000, 38000, 70000]
}

df = pd.DataFrame(data)

df.to_csv("employees.tsv", sep="\t", index=False)

print("TSV file created successfully!")
print(df)
```

# Q2.Read the TSV file into a Pandas DataFrame and display the first five rows.

# Q7.Calculate the average salary and maximum experience for each department

```python
result = df.groupby("Department").agg(
    AverageSalary=("Salary", "mean"),
    MaximumExperience=("Experience", "max")
)

print(result)
```

# Q8. Introduce at least two missing values in different columns and demonstrate two methods to handle them:
   
   - Replace missing numerical values with the column mean.
   - Remove rows containing missing values.

```python
df.loc[0, "Salary"] = None
df.loc[1, "Experience"] = None

print(df)
```

METHOD1

```python
df_mean = df.copy()

df_mean["Salary"] = df_mean["Salary"].fillna(
    df_mean["Salary"].mean()
)

df_mean["Experience"] = df_mean["Experience"].fillna(
    df_mean["Experience"].mean()
)

print(df_mean)
```

METHOD2

```python
df_drop = df.dropna()

print(df_drop)
```

# Q9.Add a new column named "AnnualSalary" by multiplying the monthly salary by 12.

```python
df_clean = df_mean.copy()

df_clean["AnnualSalary"] = df_clean["Salary"] * 12

print(df_clean)
```

# Q10.Save the cleaned DataFrame to a new TSV file without writing the index.

```python
df_clean.to_csv("cleaned_employees.tsv", sep="\t", index=False)

print("Cleaned TSV file saved successfully!")
```

# Q11. Read the cleaned TSV file again and verify its column names, dimensions, and first five rows.


```python
df_final = pd.read_csv("cleaned_employees.tsv", sep="\t")

print("Column names:")
print(df_final.columns)

print("\nDimensions:")
print(df_final.shape)

print("\nFirst five rows:")
print(df_final.head(5))
```

# Q12.Identify the department with the highest average salary and display employees whose experience is at least five years.

```python
average_salary = df_final.groupby("Department")["Salary"].mean()

highest_department = average_salary.idxmax()

print("Department with highest average salary:", highest_department)
print("Average salary:", average_salary.max())
```
